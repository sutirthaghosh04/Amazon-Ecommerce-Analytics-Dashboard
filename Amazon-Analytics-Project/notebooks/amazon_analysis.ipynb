# Data Cleaning

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df['discounted_price'] = df['discounted_price'].replace('[₹,]', '', regex=True)

df['discounted_price'] = df['discounted_price'].astype(float)
```

```python
df['discounted_price'] = df['discounted_price'].replace('[₹,]', '', regex=True)

df['discounted_price'] = df['discounted_price'].astype(float)
```

```python
df['discounted_price'] = df['discounted_price'].replace('[₹â‚¹,]', '', regex=True)

df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
```

```python
df['actual_price'] = df['actual_price'].replace('[₹â‚¹,]', '', regex=True)

df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')
```

```python
df.info()
```

```python
df['discounted_price'].unique()[:10]
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['rating_count'] = df['rating_count'].replace(',', '', regex=True)

df['rating_count'] = pd.to_numeric(df['rating_count'])
```

```python
df.info()
```

# Feature Engineering

```python
df['discount_amount'] = df['actual_price'] - df['discounted_price']
```

```python
df['estimated_revenue'] = df['discounted_price'] * df['rating_count']
```

```python
df['estimated_profit'] = df['estimated_revenue'] * 0.20
```

```python
df.head()
```

# Exploratory Data Analysis (EDA)

```python
most_reviewed = df.sort_values(by='rating_count', ascending=False)

most_reviewed[['product_name', 'rating_count']].head(10)
```

```python
top_revenue = df.sort_values(by='estimated_revenue', ascending=False)

top_revenue[['product_name', 'estimated_revenue']].head(10)
```

# Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
category_revenue.sort_values(ascending=False).head(10).plot(kind='bar')

plt.title("Top Categories by Revenue")

plt.xlabel("Category")

plt.ylabel("Revenue")

plt.xticks(rotation=90)

plt.show()
```

```python
plt.hist(df['rating'])

plt.title("Rating Distribution")

plt.xlabel("Rating")

plt.ylabel("Count")

plt.show()
```

```python
plt.scatter(df['discounted_price'], df['rating'])

plt.title("Price vs Rating")

plt.xlabel("Discounted Price")

plt.ylabel("Rating")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])

sns.heatmap(numeric_df.corr(), annot=True)

plt.show()
```
